**Optional extension only.** Kafka broker is included, but the Spark SQL Kafka source connector is **not** in the standard Spark image. Therefore this notebook demonstrates the producer and confirms events can be sent. A Spark streaming Kafka consumer requires explicitly provisioning compatible `spark-sql-kafka-0-10_2.12` dependencies (plus transitive dependencies) to driver and executors using an approved repository. Use notebook 05 for the guaranteed in-class streaming lab.


```python
from kafka import KafkaProducer
import os, json, time
producer=KafkaProducer(bootstrap_servers=os.getenv('KAFKA_BOOTSTRAP_SERVERS','kafka:9092'),value_serializer=lambda x:json.dumps(x).encode())
future=producer.send('training-payments',{'event_id':'sample-001','amount':1200000,'created_at':int(time.time())})
print('Kafka result:',future.get(timeout=30))
producer.flush();producer.close()
```

**Discussion:** Kafka offsets, checkpointing, at-least-once processing, and compatibility between Spark 3.5.x and connector artifacts.
