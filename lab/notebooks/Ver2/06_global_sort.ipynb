# Lab 06 — Global Sorting

**Tujuan:** Memahami perbedaan local sort dan global sort serta shuffle.

**Lingkungan:** Spark standalone (`spark-master:7077`), Jupyter, dua worker. Jalankan sel secara berurutan. Setiap notebook dapat dijalankan sendiri.

**Metode:** Prediksi → jalankan → amati `explain()`/Spark UI → ubah satu variabel → simpulkan.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Spark Training - Lab 06")
    .master("spark://spark-master:7077")
    .config("spark.executor.cores", "1")
    .config("spark.executor.memory", "1g")
    .config("spark.cores.max", "4")
    .config("spark.sql.shuffle.partitions", "8")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print("Spark:", spark.version)
print("Master:", spark.sparkContext.master)
```

```python
df = spark.range(0, 2_000_000, 1, 4)
print("No sorting")
df.explain("formatted")
print("Global orderBy")
ordered = df.orderBy("id", ascending=False)
ordered.explain("formatted")
```

**Prediksi:** mengapa global `orderBy` memerlukan pengaturan ulang batas range antarpartisi?

```python
ordered.show(10)
```

## Compare sortWithinPartitions

Ini bukan jaminan urutan global.

```python
local_sorted = df.sortWithinPartitions("id")
local_sorted.explain("formatted")
local_sorted.count()
```
