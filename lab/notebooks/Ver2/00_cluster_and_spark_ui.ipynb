# Lab 00 — Connect to Cluster

**Tujuan:** Mengenali driver, master, worker, executor, application, serta dashboard.

**Lingkungan:** Spark standalone (`spark-master:7077`), Jupyter, dua worker. Jalankan sel secara berurutan. Setiap notebook dapat dijalankan sendiri.

**Metode:** Prediksi → jalankan → amati `explain()`/Spark UI → ubah satu variabel → simpulkan.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Spark Training - Lab 00")
    .master("spark://spark-master:7077")
    .config("spark.executor.cores", "1")
    .config("spark.executor.memory", "1g")
    .config("spark.cores.max", "4")
    .config("spark.sql.shuffle.partitions", "8")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print("Spark:", spark.version)
print("Master:", spark.sparkContext.master)
```

## 1. Inspect Spark configuration

Buka Spark Master UI `http://localhost:8080` dan Spark Application UI `http://localhost:4040` selama notebook aktif.

```python
print("Application ID:", spark.sparkContext.applicationId)
print("Default parallelism:", spark.sparkContext.defaultParallelism)
print("Shuffle partitions:", spark.conf.get("spark.sql.shuffle.partitions"))
```

## 2. Submit a job

**Prediksi:** berapa task dan stage yang akan terlihat?

```python
numbers = spark.range(0, 100_000, 1, 4)
print("Partitions:", numbers.rdd.getNumPartitions())
print("Rows:", numbers.count())
```

## 3. Amati

Buka Spark UI → Jobs → Stages → Tasks. Catat worker/executor dan jumlah task.

**Eksperimen:** ulangi dengan `spark.range(0, 100_000, 1, 8)`; jangan asumsikan task berjalan bersamaan sebanyak jumlah partisi.

```python
numbers_8 = spark.range(0, 100_000, 1, 8)
print("Partitions:", numbers_8.rdd.getNumPartitions())
print("Rows:", numbers_8.count())
```

```python
numbers_16 = spark.range(0, 100_000, 1, 16)
print("Partitions:", numbers_16.rdd.getNumPartitions())
print("Rows:", numbers_16.count())
```
